```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv("NGSIM.csv")

# Choose variable 
variable = "follower_speed(m/s)"

# Create histogram
plt.figure(figsize=(8, 6))
plt.hist(df[variable].dropna(), bins=30)

plt.xlabel(variable)
plt.ylabel("Frequency")
plt.title(f"Distribution of {variable}")
plt.grid(True)

plt.show()
```
